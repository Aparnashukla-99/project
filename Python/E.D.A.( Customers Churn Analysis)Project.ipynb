```python
# 1. Create a crosstab with counts
ct = pd.crosstab(df['SeniorCitizen'], df['Churn'])

# 2. Convert counts to row-wise percentages
pct = ct.div(ct.sum(axis=1), axis=0) * 100

# 3. Plot stacked bar chart
plt.figure(figsize=(8,8))
bottom_vals = [0, 0]  # bottom for stacking

for churn_value in pct.columns:
    plt.bar(
        pct.index,
        pct[churn_value],
        bottom=bottom_vals,
        label=f"Churn = {churn_value}"
    )
    bottom_vals = bottom_vals + pct[churn_value].values

# 4. Label each section with % value
for i, idx in enumerate(pct.index):
    cumulative_height = 0
    for churn_value in pct.columns:
        height = pct.loc[idx, churn_value]
        plt.text(
            x=i,
            y=cumulative_height + (height / 2),
            s=f"{height:.1f}%",
            ha='center',
            va='center',
            fontsize=10
        )
        cumulative_height += height

plt.title("SeniorCitizen vs Churn (Stacked % Bar Chart)")
plt.xlabel("SeniorCitizen")
plt.ylabel("Percentage (%)")
plt.legend()
plt.show()
```

```python
# comparatively a greater percentage of people in Senior Citizen category have churned out.
```

```python
cols = [
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

plt.figure(figsize=(18, 12))

for i, col in enumerate(cols, 1):
    plt.subplot(3, 3, i)
    ax = sns.countplot(data=df, x=col, hue='Churn')

    # Add labels
    for c in ax.containers:
        ax.bar_label(c, fontsize=8)

    plt.title(col)
    plt.tight_layout()

plt.show()
```

```python
# The visualization present count plots showing customers churn(yes or No) across various servics feactures such as 'PhoneService', 'MultipleLines', 'InternetService',and 'OnlineSecurity'.
# The majority of customers who do not churn tend to have servics .
# Internetservices (particularly DSL),and OnlineSecurity enabled.
# For services like OnlineBackup ,TechSupport, and Streaming TV, churn rates are noticeably higher when these services are not used or are unavailable .
# The Charts emphasize the relationship between services usage and customers churn.
```
