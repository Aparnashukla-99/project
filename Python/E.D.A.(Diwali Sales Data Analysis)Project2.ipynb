```python
import matplotlib.pyplot as plt # work for visulaization on data.
import seaborn as sns # work for visulaization on data on the top of matplotlib as satistical.
import pandas as pd # work on series and dataframe for manupulation on data.
import numpy as np # work on array of numerical data.
```

```python
df = pd.read_csv("/content/Diwali Sales Data.csv",encoding= 'latin-1')#to avoid encoding error,use "unicode_escape".
```

```python
# drop unreleated/blank columns
df.drop(['Status','unnamed1'],axis=1,inplace=True)
```

```python
# change data type
df['Amount']=df['Amount'].astype('int')
```

```python
# rename column
df.rename(columns={'Marital_Status':'Marital_stats'})
```

```python
#E.D.A.(Exploraty Data Analysis.), Data Cleaning.
```

```python
df.columns
```

```python
ax = sns.countplot(x='Gender',data=df)
```

```python
ax = sns.countplot(x='Gender',data=df)
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
# Interpretation = Here female count is higher than male count thats shows that female purchasing power are higher than male purchasing power.
```

```python
df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
Sales_Gender=df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x ='Gender',y='Amount',data =Sales_Gender)
```

```python
# From above graphs we can see the buyers are females and even the purchasing power of females are greater than man .
# But on the top pf the above we had seen on the basis of count but here we had seen pn the basis of Amount.
```

```python
# For Age Groups with Gender.
```

```python
ax = sns.countplot(data =df,x='Age Group') # ony for age group.
```

```python
ax = sns.countplot(data =df,x='Age Group',hue='Gender')# Age Group with Genders.
```

```python
ax = sns.countplot(data =df,x='Age Group',hue='Gender')# with labels.
for bars in ax.containers:
  ax.bar_label(bars)# on the basis of count.
```

```python
# From these graph we can see that Age group between 26-35 have higher purchasing power capacity rest of other age group.
 # Among 26-35 age group between Female and male, feamle have higher purchasing power capacity.
```

```python
# Total Amount  Vs Age Group.
```

```python
sales_Age =df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)

sns.barplot(x='Age Group',y='Amount',data=sales_Age) # on the basis of Amount.
```

```python
# From above graphs we can see that most of the buyers are of age group between 26-35 yrs female.
```

```python
# State wise total number of orders.
```

```python
Sales_State= df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)

sns.set(rc={'figure.figsize':(17,5)})
sns.barplot(data = Sales_State,x='State',y='Orders')
```

```python
# From these graphs its must be shown that large no. of orderd by state is uttar pradesh, Maharashtra and karnataka and rest of other stats.
```

```python
# Total Amount of sales by from top stats (considerd as sales)
```

```python
Sales_State= df.groupby(['State'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)

sns.set(rc={'figure.figsize':(17,5)})
sns.barplot(data = Sales_State,x='State',y='Amount')
```

```python
# From above graphs we can see that unexpectedly most of the orders are from  uttar pradesh, Maharashtra and karnataka respectively but total sales amount is from uttar pradesh, Maharashtra and karnataka.
```

```python
# From Martial Status.
```

```python
ax = sns.countplot(data=df,x='Marital_Status')

sns.set(rc={'figure.figsize':(7,5)})
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
# From the above graph we can see that Martial Status = Married status have higher purchasing power than the Non-Married.
```

```python
# On the Basis of sales Amount.
```

```python
Sales_State= df.groupby(['Marital_Status','Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)

sns.set(rc={'figure.figsize':(10,5)})
sns.barplot(data = Sales_State,x='Marital_Status',y='Amount',hue='Gender')
```

```python
# From above graphs we can see that most of the buyers are married(women)and they have higher purchasing power.
```

```python
# On the basis of OCCUPATION.
```

```python
ax = sns.countplot(data=df,x='Occupation')

sns.set(rc={'figure.figsize':(27,5)})
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
Sales_State= df.groupby(['Occupation'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)

sns.set(rc={'figure.figsize':(27,5)})
sns.barplot(data = Sales_State,x='Occupation',y='Amount')
```

```python
# From above graphs we can see that most of the buyers are working in IT, Healthcare Sector and Aviation .
```

```python
# On the Basis of product Category.
```

```python
ax = sns.countplot(data=df,x='Product_Category')

sns.set(rc={'figure.figsize':(27,5)})
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
Sales_State= df.groupby(['Product_Category'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)

sns.set(rc={'figure.figsize':(40,5)})
sns.barplot(data = Sales_State,x='Product_Category',y='Amount')
```

```python
# From the above graphs we can see that most of the sold products are from food,clothing,and Electronics category.
```

```python
Sales_State= df.groupby(['Product_ID'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)

sns.set(rc={'figure.figsize':(40,5)})
sns.barplot(data = Sales_State,x='Product_ID',y='Orders')
```

```python
# Top 10 most sold products (Same thing as above)

fig1, ax1 = plt.subplots(figsize=(12,7))
df.groupby('Product_ID')['Orders'].sum().nlargest(10).sort_values(ascending=False).plot(kind='bar')
```
